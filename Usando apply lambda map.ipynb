# Compresión de Lista
Es una forma concisa y eficiente de crear listas en Python. Permite generar una nueva lista aplicando una expresión a cada elemento de una secuencia o iterable, opcionalmente filtrando elementos con una condición.

La sintaxis básica es:
[expresion for elemento in iterable if condicion]

```python
# Creando una lista de los cuadrados de los números del 1 al 10 usando list comprehension
cuadrados = [x**2 for x in range(1, 11)]
print(cuadrados)

# Creando una lista con los números pares del 1 al 100 multiplicado por 2 usando list comprehension# Creando un diccionario con los números del 1 al 10 y sus cuadrados usando dictionary comprehension
cuadrados_dict = {x: x**2 for x in range(1, 11)}
print(cuadrados_dict)
A = [i*2 for i in range(1, 101) if i % 2 == 0]
print(A)
```

# Compresión de Diccionarios
 Es una forma concisa y eficiente de crear diccionarios en Python. Permite generar un nuevo diccionario aplicando una expresión a cada elemento de una secuencia o iterable, opcionalmente filtrando elementos con una condición.

La sintaxis básica es:
{clave: valor for elemento in iterable if condicion}

```python
# Creando un diccionario con los números del 1 al 10 y sus cuadrados usando dictionary comprehension
cuadrados_dict = {x: x**2 for x in range(1, 11)}
print(cuadrados_dict)
```

# Funciones Lambda
En Python, se usa  la palabra clave __lambda__ para declarar una función anónima, por lo que nos referimos a ellas como "funciones lambda". Una función anónima se refiere a una función declarada sin nombre. Aunque sintácticamente se ven diferentes, las funciones lambda se comportan de la misma manera que las funciones regulares que se declaran usando la palabra clave __def__

Las funciones lambda se utilizan cuando se necesita una función por un período corto de tiempo.

```python
def producto(x,y):
    producto=x*y
    return(producto)

producto(5,2)
```

```python
producto = lambda x, y : x * y
producto(5,3)
```

# Map
 Se utiliza para aplicar una función a cada elemento de un iterable (como una lista o un rango) y devolver un nuevo iterable con los resultados. La sintaxis básica es:
 
map(funcion, iterable)

```python
# Usando map para crear una lista de los cuadrados de los números del 1 al 10
numeros = range(1, 11)
cuadrados = list(map(lambda x: x**2, numeros))
print(cuadrados)
```

```python
numbers = [1, 2, 3, 4]
numbers_v2 = []
for i in numbers:
  numbers_v2.append(i * 2)

numbers_v3 = list(map(lambda i: i * 2, numbers))
print(numbers_v3)
```

```python
numero_1 = [1, 2, 3, 4]
numero_2 = [5, 6, 7]

print(numero_1)
print(numero_2)
result = list(map(lambda x, y: x + y, numero_1, numero_2))
print(result)
```

```python
items = [
  {
    'product': 'camisa',
    'price': 100,
  },
  {
    'product': 'pantalones',
    'price': 300
  },
  {
    'product': 'pantalones 2',
    'price': 200
  }
]

prices = list(map(lambda item: item['price'], items))
print(items)
print(prices)

def add_taxes(item):
  item['taxes'] = item['price'] * .19
  return item

new_items = list(map(add_taxes, items))
print(new_items)
print(items)
```

# FILTER 
La función filter(), devuelve un valor que esta siendo iterado de la cual su resultado será el valor que se esta buscando en el filter.

```python
numbers = [1, 2, 3, 4, 5]
new_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(new_numbers)  # Salida: [2, 4]
```

```python
data = {'a': 5, 'b': 15, 'c': 8, 'd': 20}
filtered_data = dict(filter(lambda item: item[1] > 10, data.items()))
print(filtered_data)  # Salida: {'b': 15, 'd': 20}
```

# Reduce

La función reduce es una herramienta poderosa en Python, esencial para transformar una lista en un solo valor mediante la acumulación de sus elementos. Despuès de revisar map y filter, *reduce* completa la trilogía de funciones para la manipulación de listas. Es particularmente útil cuando necesitas realizar operaciones acumulativas como sumar todos los números en una lista o encontrar el máximo.

```python
import functools

numbers = [11, 12, 13, 14]

def accum(counter, item):
  print('counter => ',counter)
  print('item => ',item)
  return counter + item

result = functools.reduce(accum, numbers)

print(result)
```
